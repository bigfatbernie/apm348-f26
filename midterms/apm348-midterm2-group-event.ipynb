### Approximating Solutions of 1st order Systems of ODEs

Below you can find code to approximate the solution of
$$ \frac{d \vec{x}}{dt} = \vec{f}(t,\vec{x}) $$
with the **initial condition** $\vec{x}(t_0) = \vec{x}_0$.

The method will approximate the solution for the points $t_i = t_0 + i h$ for $i = 0, \ldots, n$ where the following information needs to be defined:
- `t_span`$ = ( t_0, t_n)$
- `h` where $n = \frac{t_n - t_0}{h}$ 
- `x0` corresponds to $\vec{x}_0$  is a vector
- `f(t,x)` corresponds to $\vec{f}(t,\vec{x})$ which is a vector valued function
- 

After running Euler's Method, the approximation will be stored as:
- `t` a list containing all the points $t_i$
- `x` a list containing the approximation of the solution to the system
   - `x[0]` contains a list of the values of $x_1(t_i)$
   - `x[1]` contains a list of the values of $x_2(t_i)$
   - ...

_NOTE: This code is slightly different from the practice code. It approximates the solution only until it `hits the ground`. You can comment the appropriate line of code (marked below) to obtain the full approximation._



```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

# Example usage: Solve 
#
#     dx_0/dt = x_1 
#     dx_1/dt = x_2
#     dx_2/dt = -x_0
#
#     x_0(0) = 0
#     x_1(0) = 1
#     x_2(0) = 0
#
# from t=0 to t=5
#
# Note: Function names start with x_0 (not x_1).



######### BEGIN STUDENT INPUT #########


######### PARAMETER SETS FOR THE GROUP TEST QUESTIONS #########
# Comment the one you don't want to use

A = 70
B = 3
C = 0.15

W0 = 1
H = 8
s = 1.5

k = 0.12

## PARAMETERS FOR Q1

T0 = 6

## PARAMETERS FOR Q2

# T0 = 9

######### OTHER INPUT #########

def f(t, x):
    return [
        x[1],   # = dx[0]/dt
        x[2],   # = dx[1]/dt        
        -x[0]   # = dx[2]/dt
    ]

x0 = [0,1,0]

h = 0.01  # Step size    
t_span = (0, 5) # Interval for solution approximation

######### END OF STUDENT INPUT #########


def hit_ground(t, x):
    r, h, v = x
    return h

hit_ground.terminal = True
hit_ground.direction = -1


t_eval = np.arange(t_span[0],t_span[1]+h,h)
    
sol = solve_ivp(
    f,
    t_span,
    x0,
    t_eval=t_eval,
    events=hit_ground,       # Approximation will stop when the 2nd variable `h` goes from positive to negative (solutions hits the ground)
                             # Comment this line if you want to see the full approximation in `t_span`
    dense_output=True
)

t = t_eval
x = sol.sol(t)
```

### Graph one variable in component space

```python

# We want to graph variable:
i = 1    # to plot x[i]


"""Plots the solution obtained from Euler's method."""
plt.plot(t, x[i], linestyle='-')
plt.xlabel('t')
plt.ylabel(f'x_{i}')
plt.grid(True)
# plt.legend()
plt.show()
```

### Graph two variables in phase space

```python

# We want to graph variables x[i] vs x[j] 
i = 1      # variable that will take values on the x-axis
j = 2      # variable that will take values on the y-axis


"""Plots the solution obtained from Euler's method."""
plt.plot(x[i], x[j], linestyle='-')
plt.xlabel(f'x_{i}')
plt.ylabel(f'x_{j}')
plt.grid(True)
# plt.legend()
plt.show()
```
